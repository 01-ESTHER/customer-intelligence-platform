# 06 | Repeat-Purchase Model Improvement

## Objective

Improve the reliability, customer coverage and predictive
performance of the existing 60-day repeat-purchase model.

## Baseline

Unweighted Logistic Regression using nine first-purchase
features.

### Established validation performance

- Average Precision: 0.0375
- ROC AUC: 0.6305
- Brier Score: 0.0189
- Log Loss: 0.0938

### Previously observed limitations

- Missing features exclude some eligible customers.
- Predictive discrimination weakens across historical periods.
- Predicted repeat-purchase probabilities are overestimated.
- The current model may not capture complex feature interactions.

## Development Plan

1. Load the existing feature dataset and baseline model.
2. Recreate the chronological development populations.
3. Introduce appropriate missing-value handling.
4. Train and compare alternative model configurations.
5. Assess ranking performance and probability calibration.
6. Select the preferred development model using validation data.
7. Document improvements, limitations and next steps.

## Evaluation Principles

- Preserve the original Logistic Regression baseline.
- Fit preprocessing transformations using training data only.
- Compare models on matched validation populations.
- Do not use the reserved holdout for model selection.
- Treat model explanations as associations, not causal effects.
